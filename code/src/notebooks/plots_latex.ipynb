```python
import torch
from torchvision import datasets, transforms
import matplotlib.pyplot as plt
import os

# Path to your local FashionMNIST data
data_path = r"C:\Users\marou\Desktop\Uni\HPC\Projekt\code\jobs\model_training_benchmark\data"

# Define a transform to convert PIL images to tensors
transform = transforms.ToTensor()

# Load the training dataset
train_dataset = datasets.FashionMNIST(root=data_path, train=True, download=False, transform=transform)

# Labels for FashionMNIST
classes = ['T-Shirt/Top', 'Trouser', 'Pullover', 'Dress', 'Coat', 
           'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle Boot']

# Dictionary to store one example per class
examples = {}

# Iterate through dataset to get one example per class
for img, label in train_dataset:
    if label not in examples:
        examples[label] = img
    if len(examples) == 10:
        break

# Plot all classes in a single figure
plt.figure(figsize=(12, 4))
for i in range(10):
    plt.subplot(2, 5, i+1)
    plt.imshow(examples[i].squeeze(), cmap='gray')
    plt.title(classes[i])
    plt.axis('off')

# Save the figure
output_dir = os.path.join(data_path, 'plots')
#os.makedirs(output_dir, exist_ok=True)
#plt.savefig(os.path.join(output_dir, 'fashionmnist_classes.png'), bbox_inches='tight')
plt.show()
len(train_dataset)
```
